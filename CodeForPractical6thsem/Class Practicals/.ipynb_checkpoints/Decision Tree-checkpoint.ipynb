```python
import pandas as pd
df = pd.read_excel(r"C:\Users\DELL\OneDrive\Documents\Semester 6 Lab Manual\Machine Learning\Decision tree.xlsx")
df
```

```python
x=df[['company','job','degree']]
x
```

```python
y=df['salery']
y
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
x['comn'] = LabelEncoder().fit_transform(x['company'])
x['jobn'] = LabelEncoder().fit_transform(x['job'])
x['degreen'] = LabelEncoder().fit_transform(x['degree'])
x = x[['comn','jobn','degreen']]
x
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(x,y)
```

```python
model.predict([[0,0,1]])
```
